```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import tree
```

```python
caminho = r'/content/drive/MyDrive/Colab Notebooks/Dados/vendas.xlsx'
```

```python
dados = pd.read_excel(caminho)
dados.head()
```

```python
x = dados[['media_rating',"media_pedidos","test"]]
y = dados[['cliente_comprou']]
```

```python
x = pd.get_dummies(x)
x.head()
```

```python
arvore = tree.DecisionTreeClassifier()
modelo = arvore.fit(x,y)
```

```python
modelo
```

```python
fig = plt.figure(figsize=(40,40))
tree = tree.plot_tree(modelo, feature_names=x.columns, filled=True, class_names=['Yes','No'])
```
